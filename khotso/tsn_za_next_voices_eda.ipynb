```python
import numpy as np

thresholds = [5, 10, 15, 30, 60]  # edit/add thresholds you want
total_count = len(df)
total_dur = df['duration'].sum()

print(f"Total recordings: {total_count}, total duration: {total_dur:.2f}s ({total_dur/3600:.2f}h)\n")

# Cumulative: count of recordings shorter than each threshold
print("Cumulative ( < T seconds ):")
for t in thresholds:
    cnt = (df['duration'] < t).sum()
    pct = cnt / total_count if total_count else 0
    dur = df.loc[df['duration'] < t, 'duration'].sum()
    print(f" < {t:>2}s : {cnt:>6} ({pct:.2%}), duration = {dur:.2f}s ({dur/3600:.4f}h)")

# Bucketed: e.g. 0-5,5-10,10-15...
edges = [0] + thresholds
print("\nBuckets (low <= dur < high):")
for low, high in zip(edges, edges[1:]):
    mask = (df['duration'] >= low) & (df['duration'] < high)
    cnt = mask.sum()
    pct = cnt / total_count if total_count else 0
    dur = df.loc[mask, 'duration'].sum()
    print(f" {low:>2}-{high:>2}s: {cnt:>6} ({pct:.2%}), duration = {dur:.2f}s")

# Optional last bucket (>= last threshold)
low = thresholds[-1]
mask = df['duration'] >= low
cnt = mask.sum()
pct = cnt / total_count if total_count else 0
dur = df.loc[mask, 'duration'].sum()
print(f" >={low:>2}s : {cnt:>6} ({pct:.2%}), duration = {dur:.2f}s")
```

```python
#count none or empty transcripts
empty_transcripts = df['transcript'].isnull().sum() + (df['transcript'].str.strip() == '').sum()
print(f"Number of empty or null transcripts: {empty_transcripts}")
```

## Word analysis

```python
df['transcript_word_count'] = df['transcript'].apply(lambda x: len(str(x).split()))
plt.hist(df['transcript_word_count'], bins=20, color="lightgreen", edgecolor="black")
plt.title("Distribution of Transcript Word Counts")
plt.xlabel("Word Count")
plt.ylabel("Frequency")
plt.show()
```

```python
#avereage word count per transcript
average_word_count = df['transcript_word_count'].mean()
print(f"Average word count per transcript: {average_word_count:.2f} words")
```

```python
from collections import Counter
all_words = ' '.join(df['transcript'].dropna()).lower().split()
word_counts = Counter(all_words)
top_words = word_counts.most_common(20)
print(f"total unique words: {len(word_counts)}")
```

```python
for i, (word, count) in enumerate(top_words, start=1):
    print(f"{i}. {word}: {count}")
```

## Character analysis

```python
df['transcript_char_count'] = df['transcript'].apply(lambda x: len(str(x)))

plt.hist(df['transcript_char_count'], bins=20, color="lightgreen", edgecolor="black")
plt.title(f"Distribution of Transcript Character Counts")
plt.xlabel("Character Count")
plt.ylabel("Frequency")
plt.show()

```

```python
#average character count per transcript
average_char_count = df['transcript_char_count'].mean()
print(f"Average character count per transcript: {average_char_count:.2f} characters")
```

Top characters in transcript

```python
from collections import Counter
all_chars = ''.join(df['transcript'].dropna()).lower()
filtered_chars = [c for c in all_chars if not c.isspace()]
char_counts = Counter(filtered_chars)
top_chars = char_counts.most_common(20)

number_of_unique_chars = len(char_counts)
print(f" Number of Unique characters: {number_of_unique_chars}")


```
