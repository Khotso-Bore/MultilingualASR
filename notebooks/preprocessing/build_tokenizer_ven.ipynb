## Verify: lossless round-trip on Tshivenda diacritics

```python
from transformers import Wav2Vec2CTCTokenizer
tok = Wav2Vec2CTCTokenizer.from_pretrained("../tokenizers/ven")

tests = ["ndi khou ṱoḓa u ḓivha", "ḽa mvelele ṅwaha ṋamusi"]
for t in tests:
    back = tok.decode(tok(t).input_ids)
    print(f"{'OK ' if back == t else 'FAIL'} {t!r} -> {back!r}")
print("vocab size:", tok.vocab_size)
```
