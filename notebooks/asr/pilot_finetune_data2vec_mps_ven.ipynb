# Tshivenda data2vec-audio Pilot Fine-Tune (M4 / MPS)

Wraps `src/asr/pilot_finetune_data2vec_mps_ven.py` - the sixth ASR model family attempt.
AfriHuBERT, MMS, and w2v-BERT all collapsed into predicting a single dominant token
(see `notes/pilot-ven-results.md`) - all three pretrain on some form of discretized
target. `facebook/data2vec-audio-large` regresses onto continuous teacher
representations instead, a genuinely different objective, directly testing whether
discretization is the common thread. Honest caveat: English-only pretraining
(Librispeech), the weakest multilingual transfer prior tried so far.

NOT the real Stage 1 run (needs a CUDA GPU) - this is for quick local iteration and
sanity checks only.

Needs `PYTORCH_ENABLE_MPS_FALLBACK=1` set (CTC loss has no MPS kernel) - set it
as an environment variable before launching Jupyter, or via the cell below.

**Kernel**: MultilingualASR.

```python
import os
os.environ["PYTORCH_ENABLE_MPS_FALLBACK"] = "1"
```
