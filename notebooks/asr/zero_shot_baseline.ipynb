# Tshivenda Zero-Shot ASR Baseline

Runs pretrained Whisper (no fine-tuning) on the preprocessed Tshivenda test sets and
scores WER/CER with jiwer - the "before" numbers for Sub-question 1. Logic lives in
`src/asr/zero_shot_baseline.py`.

Notes:
- Whisper has no Tshivenda language token, so it auto-detects (and usually
  mis-detects) the language. Very high WER is the expected zero-shot condition.
- Wav2Vec2 XLS-R has no Tshivenda CTC head, so it has no zero-shot mode -
  this baseline is Whisper-only by design.
- Predictions are normalised with `src/text_norm.py` before scoring.

**Kernel**: MultilingualASR. Debug with `openai/whisper-small`; use
`openai/whisper-large-v3` for the reported numbers (slow locally - see summary).

```python
import sys
sys.path.insert(0, "../../src/asr")

from zero_shot_baseline import evaluate

MODEL = "openai/whisper-small"   # switch to "openai/whisper-large-v3" for reported numbers
LIMIT = 20                        # clips per corpus; 200+ for the real baseline
```

```python
results = evaluate(MODEL, limit=LIMIT, seed=42, n_examples=3)
```
