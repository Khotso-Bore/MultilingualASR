## Verify: balance, leakage, length confound, diacritics

```python
import pandas as pd

df = pd.read_csv(OUTPUT_CSV)
print("label balance:\n", df["label"].value_counts())

leak = df.groupby("source_id")["split"].nunique()
print("\nsource_ids with real/fake split mismatch (should be 0):", (leak > 1).sum())

wc = df["text"].str.split().str.len()
print("\nmean word count - real:", wc[df.label==1].mean(), "| fake:", wc[df.label==0].mean())

DIA = set("ḓḽṅṋṱ")
print("\nrows with tshivenda diacritics:", df["text"].apply(lambda t: bool(set(t) & DIA)).sum(), "/", len(df))
```
