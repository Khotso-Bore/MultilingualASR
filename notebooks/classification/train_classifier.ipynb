# Tshivenda Misinformation Classifier - Local (M4)

Wraps `src/train_classifier.py` - fine-tunes and grouped-5-fold-cross-validates
the classifier locally. Small enough to run on CPU/MPS in well under an hour
(unlike ASR training) - the Colab version (`colab_classifier_ven.ipynb`) is
only needed for speed, not because this can't run locally.

See `notes/tshivenda-classifier-proxy.md` for the dataset's rationale and
limitations before citing these numbers.

**Kernel**: MultilingualASR.

## AfroXLM-RoBERTa (primary)

```python
import sys
sys.path.insert(0, "../../src/classification")
from train_classifier import cross_validate

afroxlmr_accs, afroxlmr_f1s = cross_validate(
    model_name="Davlan/afro-xlmr-base", folds=5, epochs=15, batch_size=4,
    seed=42, learning_rate=1e-3, freeze_base=True)
```

## XLM-RoBERTa (comparison baseline)

```python
xlmr_accs, xlmr_f1s = cross_validate(
    model_name="xlm-roberta-base", folds=5, epochs=15, batch_size=4,
    seed=42, learning_rate=1e-3, freeze_base=True)
```

## Compare

```python
import numpy as np
print(f"{'model':<20} {'accuracy':>18} {'macro F1':>18}")
print(f"{'afro-xlmr-base':<20} {np.mean(afroxlmr_accs):.3f} +/- {np.std(afroxlmr_accs):.3f}"
      f"   {np.mean(afroxlmr_f1s):.3f} +/- {np.std(afroxlmr_f1s):.3f}")
print(f"{'xlm-roberta-base':<20} {np.mean(xlmr_accs):.3f} +/- {np.std(xlmr_accs):.3f}"
      f"   {np.mean(xlmr_f1s):.3f} +/- {np.std(xlmr_f1s):.3f}")
```
